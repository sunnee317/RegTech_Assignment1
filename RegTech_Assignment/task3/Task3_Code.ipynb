## 3. PSI-Based Distribution Drift

The previous drift checks compare average values and rates. To make the distribution-shift test more formal, this section calculates Population Stability Index (PSI). A common monitoring interpretation is: PSI below 0.10 = low drift, 0.10 to 0.25 = moderate drift, and above 0.25 = significant drift requiring model review.


```python
def psi_numeric(expected, actual, buckets=10):
    """Calculate PSI for numeric variables using training quantile bins."""
    expected = pd.Series(expected).dropna().astype(float)
    actual = pd.Series(actual).dropna().astype(float)

    quantiles = np.linspace(0, 1, buckets + 1)
    breakpoints = np.unique(np.quantile(expected, quantiles))

    if len(breakpoints) <= 2:
        breakpoints = np.linspace(expected.min(), expected.max(), buckets + 1)

    breakpoints[0] = -np.inf
    breakpoints[-1] = np.inf

    expected_counts = pd.cut(expected, bins=breakpoints, include_lowest=True).value_counts(sort=False)
    actual_counts = pd.cut(actual, bins=breakpoints, include_lowest=True).value_counts(sort=False)

    expected_pct = expected_counts / expected_counts.sum()
    actual_pct = actual_counts / actual_counts.sum()

    expected_pct = expected_pct.replace(0, 0.0001)
    actual_pct = actual_pct.replace(0, 0.0001)

    return float(((actual_pct - expected_pct) * np.log(actual_pct / expected_pct)).sum())


def psi_categorical(expected, actual):
    """Calculate PSI for categorical variables."""
    expected = pd.Series(expected).fillna("Missing").astype(str)
    actual = pd.Series(actual).fillna("Missing").astype(str)
    categories = sorted(set(expected.unique()).union(set(actual.unique())))

    expected_pct = expected.value_counts(normalize=True).reindex(categories, fill_value=0).replace(0, 0.0001)
    actual_pct = actual.value_counts(normalize=True).reindex(categories, fill_value=0).replace(0, 0.0001)

    return float(((actual_pct - expected_pct) * np.log(actual_pct / expected_pct)).sum())


psi_results = []

numeric_psi_cols = [
    "income_sgd",
    "loan_amount_sgd",
    "debt_to_income",
    "model_score",
    "credit_history_months"
]

categorical_psi_cols = [
    "employment_status",
    "gender",
    "ethnicity",
    "postal_district"
]

for col in numeric_psi_cols:
    psi_results.append({
        "Variable": col,
        "PSI": psi_numeric(train[col], live[col])
    })

for col in categorical_psi_cols:
    psi_results.append({
        "Variable": col,
        "PSI": psi_categorical(train[col], live[col])
    })

psi_table = pd.DataFrame(psi_results).sort_values("PSI", ascending=False)

def classify_psi(value):
    if value >= 0.25:
        return "Significant drift"
    if value >= 0.10:
        return "Moderate drift"
    return "Low / no material drift"

psi_table["Interpretation"] = psi_table["PSI"].apply(classify_psi)
psi_table["PSI"] = psi_table["PSI"].round(4)

display(psi_table)
```

## 4. Model Performance Monitoring: AUC

The model score is designed so that higher scores indicate better credit quality. AUC is therefore calculated using the non-default outcome as the positive class. If live AUC falls, the model's ability to separate lower-risk from higher-risk applicants has weakened.


```python
train_auc = roc_auc_score(1 - train["actual_default"], train["model_score"])
live_auc = roc_auc_score(1 - live["actual_default"], live["model_score"])
auc_change = live_auc - train_auc

auc_table = pd.DataFrame({
    "Metric": ["AUC"],
    "Training": [train_auc],
    "Live": [live_auc],
    "Change": [auc_change]
})

auc_table[["Training", "Live", "Change"]] = auc_table[["Training", "Live", "Change"]].round(4)
display(auc_table)

if auc_change < -0.01:
    print(f"PERFORMANCE ALERT: AUC declined by {abs(auc_change):.4f}. Model validation review is recommended.")
else:
    print(f"AUC change is {auc_change:.4f}. Continue routine monitoring.")
```

## 5. Fairness and Subgroup Monitoring

This section compares approval outcomes across gender, ethnicity, and employment status groups. Gender and ethnicity are especially relevant to fair-lending monitoring. Employment status is not treated as a protected-class breach by itself, but the large approval gap should still be reviewed for adverse action explanation quality and possible proxy-risk concerns.


```python
def group_monitoring_table(data, group_col):
    table = data.groupby(group_col).agg(
        applicant_count=("application_id", "count"),
        approval_rate=("approved", "mean"),
        default_rate=("actual_default", "mean"),
        avg_model_score=("model_score", "mean"),
        avg_dti=("debt_to_income", "mean")
    ).reset_index()

    return table.sort_values(group_col)


def format_monitoring_table(table):
    formatted = table.copy()
    formatted["approval_rate"] = (formatted["approval_rate"] * 100).round(2).astype(str) + "%"
    formatted["default_rate"] = (formatted["default_rate"] * 100).round(2).astype(str) + "%"
    formatted["avg_model_score"] = formatted["avg_model_score"].round(3)
    formatted["avg_dti"] = formatted["avg_dti"].round(3)
    return formatted


def approval_disparity(data, group_col):
    rates = data.groupby(group_col)["approved"].mean()
    min_rate = rates.min()
    max_rate = rates.max()
    disparity_ratio = min_rate / max_rate
    return rates, disparity_ratio
```

```python
gender_train = group_monitoring_table(train, "gender")
gender_live = group_monitoring_table(live, "gender")
ethnicity_train = group_monitoring_table(train, "ethnicity")
ethnicity_live = group_monitoring_table(live, "ethnicity")
employment_train = group_monitoring_table(train, "employment_status")
employment_live = group_monitoring_table(live, "employment_status")

print("Live Data - Gender Monitoring")
display(format_monitoring_table(gender_live))

print("Live Data - Ethnicity Monitoring")
display(format_monitoring_table(ethnicity_live))

print("Live Data - Employment Status Monitoring")
display(format_monitoring_table(employment_live))
```

```python
gender_rates, gender_disparity_ratio = approval_disparity(live, "gender")
ethnicity_rates, ethnicity_disparity_ratio = approval_disparity(live, "ethnicity")
employment_rates, employment_disparity_ratio = approval_disparity(live, "employment_status")

print("Gender approval rates:")
print(gender_rates)
print(f"Gender disparity ratio: {gender_disparity_ratio:.3f}")

print("\nEthnicity approval rates:")
print(ethnicity_rates)
print(f"Ethnicity disparity ratio: {ethnicity_disparity_ratio:.3f}")

print("\nEmployment status approval rates:")
print(employment_rates)
print(f"Employment status disparity ratio: {employment_disparity_ratio:.3f}")
```

## 6. Jurisdiction-Specific Rule Engine

The same monitoring evidence is interpreted differently by different regulators. This prototype separates OCC-style fair-lending and adverse-action review from MAS-style governance escalation and model lifecycle monitoring.


```python
model_score_psi = float(psi_table.loc[psi_table["Variable"] == "model_score", "PSI"].iloc[0])
employment_psi = float(psi_table.loc[psi_table["Variable"] == "employment_status", "PSI"].iloc[0])

monitoring_indicators = {
    "default_rate_change_pp": (live["actual_default"].mean() - train["actual_default"].mean()) * 100,
    "model_score_change": live["model_score"].mean() - train["model_score"].mean(),
    "low_score_share_change_pp": ((live["model_score"] < 0.50).mean() - (train["model_score"] < 0.50).mean()) * 100,
    "median_income_change_pct": ((live["income_sgd"].median() - train["income_sgd"].median()) / train["income_sgd"].median()) * 100,
    "gender_disparity_ratio": gender_disparity_ratio,
    "ethnicity_disparity_ratio": ethnicity_disparity_ratio,
    "employment_disparity_ratio": employment_disparity_ratio,
    "model_score_psi": model_score_psi,
    "employment_status_psi": employment_psi,
    "train_auc": train_auc,
    "live_auc": live_auc,
    "auc_change": auc_change
}

monitoring_indicators
```

```python
def occ_compliance_check(indicators):
    alerts = []
    
    if indicators["gender_disparity_ratio"] < 0.80:
        alerts.append("Gender approval disparity may require fair lending review.")
    
    if indicators["ethnicity_disparity_ratio"] < 0.80:
        alerts.append("Ethnicity approval disparity may require fair lending review.")
    
    if indicators["employment_disparity_ratio"] < 0.80:
        alerts.append("Employment-status approval gap should be reviewed for adverse action explanation quality and proxy risk.")
    
    if indicators["default_rate_change_pp"] > 5:
        alerts.append("Default rate increased materially. Model validation should be reviewed.")
    
    if indicators["auc_change"] < -0.01:
        alerts.append("AUC declined. Score separation has weakened and validation evidence should be updated.")
    
    if indicators["model_score_change"] < -0.05:
        alerts.append("Average model score declined. Adverse action explanations may need closer review.")
    
    if not alerts:
        alerts.append("No major OCC-style alert triggered under the simplified thresholds.")
    
    return alerts


def mas_compliance_check(indicators):
    alerts = []
    
    if indicators["model_score_psi"] > 0.25:
        alerts.append("Model score PSI exceeds 0.25. Significant model drift requires governance escalation.")
    
    if indicators["default_rate_change_pp"] > 5:
        alerts.append("Default rate drift exceeds monitoring threshold. Governance review is recommended.")
    
    if indicators["low_score_share_change_pp"] > 10:
        alerts.append("Low-score applicant share increased materially. Model drift review is recommended.")
    
    if indicators["median_income_change_pct"] < -10:
        alerts.append("Median income declined materially. Deployment environment has shifted.")
    
    if indicators["auc_change"] < -0.01:
        alerts.append("AUC deterioration indicates that post-deployment validation should be refreshed.")
    
    if indicators["gender_disparity_ratio"] < 0.80 or indicators["ethnicity_disparity_ratio"] < 0.80:
        alerts.append("Fairness indicator requires review under responsible AI governance.")
    
    if not alerts:
        alerts.append("No major MAS-style governance alert triggered under the simplified thresholds.")
    
    return alerts
```

```python
print("=== OCC Compliance Output ===")
for alert in occ_compliance_check(monitoring_indicators):
    print("-", alert)

print("\n=== MAS Compliance Output ===")
for alert in mas_compliance_check(monitoring_indicators):
    print("-", alert)
```

## 7. Sensitivity Analysis

This simple sensitivity check tests whether the compliance conclusion changes if the live default rate becomes even higher. It shows that once the default-rate increase exceeds the monitoring threshold, both jurisdictions continue to require review, although the reason for review is framed differently.


```python
base_default_change = monitoring_indicators["default_rate_change_pp"]

sensitivity_results = []

for extra_increase in [0, 2, 5, 8, 10]:
    simulated_default_change = base_default_change + extra_increase
    
    sensitivity_results.append({
        "Additional Default Rate Increase (pp)": extra_increase,
        "Simulated Default Rate Change (pp)": round(simulated_default_change, 2),
        "OCC Response": "Model validation / adverse action review" if simulated_default_change > 5 else "Routine monitoring",
        "MAS Response": "Governance escalation" if simulated_default_change > 5 else "Routine monitoring"
    })

sensitivity_table = pd.DataFrame(sensitivity_results)
display(sensitivity_table)
```
